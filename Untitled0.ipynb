```python
import pandas as pd

# A direct public URL to a standard Fake News dataset matching the Kaggle layout
url = "https://raw.githubusercontent.com/raj1603chdry/Fake-News-Detection-System/master/datasets/train.csv"

print("Downloading and loading the dataset... Please wait.")
data = pd.read_csv(url)
print("Data successfully loaded!")

# Let's peek at the first 5 rows of our data
data.head()
```

```python
import re

# 1. Assign features and targets matching your column names
X = data['news'].astype(str)

# Convert True/False into numeric 1/0 for the ML algorithms
y = data['label'].map({True: 1, False: 0})

# 2. Text Cleaning Function (Required by your IICT workflow)
def clean_text(text):
    text = re.sub(r'\W', ' ', text)         # Removes punctuation and special characters
    text = text.lower()                     # Converts all text to lowercase
    text = re.sub(r'\s+', ' ', text).strip() # Removes any accidental double spaces
    return text

# Apply the function to your text data
print("Cleaning text data... Please wait.")
X_clean = X.apply(clean_text)
print("Text cleaning complete!")

# Let's check a before-and-after comparison of the first article
print("\n--- Original Text Sample ---")
print(X.iloc[0][:120] + "...")

print("\n--- Cleaned Text Sample ---")
print(X_clean.iloc[0][:120] + "...")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split

# 1. Initialize the TF-IDF Vectorizer
# We limit it to the top 5,000 most important words to keep execution fast and clean
vectorizer = TfidfVectorizer(max_features=5000)

print("Converting text to numbers using TF-IDF...")
X_vec = vectorizer.fit_transform(X_clean)
print("Text vectorized successfully!")

# 2. Split the dataset into Training (80%) and Testing (20%) sets
X_train, X_test, y_train, y_test = train_test_split(X_vec, y, test_size=0.2, random_state=42)

print("\n--- Dataset Split Summary ---")
print(f"Total training samples: {X_train.shape[0]}")
print(f"Total testing samples: {X_test.shape[0]}")
print(f"Number of unique word features: {X_train.shape[1]}")
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, classification_report

# 1. Initialize all 4 models required by the IICT framework
models = {
    "KNN (Non-Parametric)": KNeighborsClassifier(n_neighbors=5),
    "Logistic Regression (Parametric)": LogisticRegression(max_iter=1000),
    "Random Forest (Ensemble)": RandomForestClassifier(n_estimators=100, random_state=42),
    "Neural Network (Deep Learning)": MLPClassifier(hidden_layer_sizes=(100,), max_iter=300, random_state=42)
}

# 2. Loop through each model, train it, and print metrics
for name, model in models.items():
    print(f"\n🚀 Training {name}...")

    # Train the model on training data
    model.fit(X_train, y_train)

    # Make predictions on the unseen test data
    preds = model.predict(X_test)

    # Output the required evaluation metrics
    print(f"✅ {name} Performance:")
    print("Accuracy:", accuracy_score(y_test, preds))
    print(classification_report(y_test, preds, target_names=['Real News', 'Fake News']))
    print("-" * 60)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix

# We will grab the Logistic Regression model from our trained dictionary to plot
lr_model = models["Logistic Regression (Parametric)"]
lr_preds = lr_model.predict(X_test)

# Compute the matrix values
cm = confusion_matrix(y_test, lr_preds)

# Plot the matrix visually using a heatmap
plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Real News', 'Fake News'],
            yticklabels=['Real News', 'Fake News'])

plt.title('Confusion Matrix: Logistic Regression Classifier')
plt.ylabel('Actual Truth')
plt.xlabel('AI Predicted Label')
plt.show()
```
